```python
import json
from discovery_child_development import PROJECT_DIR
```

```python
from discovery_child_development.utils import openalex_utils
```

```python
import pandas as pd
```

```python
for i in [1, 2, 3, 4, 5, 6]:
    data = json.load(open(PROJECT_DIR / f'inputs/data/openalex_keywords/json/output_{i}.json'))
    data_df = pd.DataFrame(data)

    _data_df = (
        data_df[['id', 'publication_year', 'title', 'abstract_inverted_index', 'authorships', 'language']]
        .dropna(subset=['abstract_inverted_index'])
        .dropna(subset=['title'])
        .query("language == 'en'")
        .assign(abstract = lambda df: df.abstract_inverted_index.apply(openalex_utils.deinvert_abstract))
        .drop(columns=['abstract_inverted_index'])
    )

    _data_df.to_parquet(PROJECT_DIR / f'inputs/data/openalex_keywords/parquet/output_{i}.parquet', index=False)
```
