## Processing openalex json downloads

```python
import json
from discovery_child_development import PROJECT_DIR
```

```python
from discovery_child_development.utils import openalex_utils
```

```python
import pandas as pd
```

```python
for i in [1, 2, 3, 4, 5, 6]:
    data = json.load(open(PROJECT_DIR / f'inputs/data/openalex_keywords/json/output_{i}.json'))
    data_df = pd.DataFrame(data)

    _data_df = (
        data_df[['id', 'publication_year', 'title', 'abstract_inverted_index', 'authorships', 'language']]
        .dropna(subset=['abstract_inverted_index'])
        .dropna(subset=['title'])
        .query("language == 'en'")
        .assign(abstract = lambda df: df.abstract_inverted_index.apply(openalex_utils.deinvert_abstract))
        .drop(columns=['abstract_inverted_index'])
    )

    _data_df.to_parquet(PROJECT_DIR / f'inputs/data/openalex_keywords/parquet/output_{i}.parquet', index=False)
```

## Checking openalex data

```python
import pandas as pd
from nesta_ds_utils.loading_saving import S3
from discovery_child_development import S3_BUCKET, PROJECT_DIR

OUTPUT_FILENAME = "keywords_31May2024"
ENRICHED_DATA_DIR = PROJECT_DIR / "outputs/enrichments"
```

```python
dfs = []
for i in list(range(0,7)):
    dfs.append(
        S3.download_obj(
            S3_BUCKET,
            path_from = f"data/outputs/binary_classifier/{OUTPUT_FILENAME}_{i}.parquet",
            download_as="dataframe",
        )
        .query("predictions == 1")
        .assign(source = "openalex")
    )

print(len(dfs))
dfs = pd.concat(dfs)
dfs.to_csv(ENRICHED_DATA_DIR / "openalex_relevance_labels_only_relevant.csv", index=False)
```
