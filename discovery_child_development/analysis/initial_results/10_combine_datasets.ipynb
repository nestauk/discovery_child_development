# Combine datasets
- combine tables for Atlas
- combine trends tables for plots

```python
from discovery_child_development import PROJECT_DIR
import pandas as pd
import ast
```

## Raw data for Atlas

```python
gtr_df = pd.read_csv(PROJECT_DIR / "outputs/data/tables/gtr_final.csv").assign(dataset = 'ukri', country_code='GB')
```

```python
openalex_df = (
    pd.read_csv(PROJECT_DIR / "outputs/data/tables/openalex_final.csv")
)
```

```python
patents_df = pd.read_csv(PROJECT_DIR / "outputs/data/tables/patents_final.csv")
```

```python
combined_df = pd.concat([
    gtr_df,
    openalex_df,
    patents_df,
], ignore_index=True)[['dataset', 'id', 'url', 'text', 'year', 'country_code', 'topic', 'minor_category', 'major_category', 'topic_code', 'amount', 'lead_funder']]
```

```python
combined_df.to_csv(PROJECT_DIR / "outputs/data/tables/combined_gtr_openalex_patents_final.csv", index=False)
```

## Data for detailed application plots

```python
import utils
topics_df = utils.load_topic_data()
```

```python
# applications_gtr_df = pd.read_csv(PROJECT_DIR / 'outputs/data/tables/tech_applications_gtr_amount.csv')
# applications_crunchbase_df = pd.read_csv(PROJECT_DIR / 'outputs/data/tables/tech_applications_crunchbase_amount.csv')

applications_gtr_df = pd.read_csv(PROJECT_DIR / 'outputs/data/tables/tech_applications_gtr_counts.csv')
applications_crunchbase_df = pd.read_csv(PROJECT_DIR / 'outputs/data/tables/tech_applications_crunchbase_counts.csv')

applications_openalex_df = pd.read_csv(PROJECT_DIR / 'outputs/data/tables/tech_applications_openalex.csv')
applications_patents_df = pd.read_csv(PROJECT_DIR / 'outputs/data/tables/tech_applications_patents.csv')
```

```python
base_df = (
    applications_openalex_df
    .merge(topics_df[['type', 'name']], on='name')
    .sort_values(['type', 'Total'], ascending=[True, False])
)[['type', 'name']]

applications_gtr_df = (
    base_df
    .merge(applications_gtr_df, on='name', how='left').fillna(0)
)

applications_crunchbase_df = (
    base_df
    .merge(applications_crunchbase_df, on='name', how='left').fillna(0)
)

applications_openalex_df = (
    base_df
    .merge(applications_openalex_df, on='name', how='left').fillna(0)
)

applications_patents_df = (
    base_df
    .merge(applications_patents_df, on='name', how='left').fillna(0)
)
```
